```python
# Q22
a = "hello"
b = "".join(["he", "llo"])
print(a == b)
print(a is b)
```

Explanation:
a == b
== checks value equality — whether the contents of the two objects are the same.
"hello" and "".join(["he", "llo"]) both produce the string "hello".
Since their contents match, a == b returns True.
a is b
is checks object identity — whether both variables point to the same object in memory.
a is assigned directly from a string literal "hello".
In Python, string literals may be interned (stored in a shared pool for efficiency).
b is created at runtime using "".join(...), which produces a new string object with the same content but a different memory address.
Therefore, a and b are different objects, so a is b returns False.

```python
# Q35
import copy

a = [[1, 2], [3, 4]]
b = copy.copy(a)
b[0][1] = 99
print(a)
```

Explanation:
copy.copy() makes a shallow copy — it copies the outer list, but inner lists are still shared between a and b.
When you change b[0][1], you’re modifying the same inner list that a[0] points to.
So a is also affected.

```python
# Q36
import copy

a = [[1, 2], [3, 4]]
b = copy.deepcopy(a)
b[0][1] = 99
print(a)
```

Explanation: deepcopy duplicates all nested objects, so b has its own separate inner lists. Modifying b does not affect a.
If you had used copy.copy() (shallow copy), changes to inner lists would also affect a.

```python
# Q37
import copy

a = [[10], [20]]
b = a.copy()
a[1].append(30)
print(b)
```

Explanation:
list.copy() makes a shallow copy — it copies the outer list but references the same inner lists.
So when you append 30 to a[1], b[1] changes too because both point to the same inner list.

```python
# Q38
import copy

a = [[10], [20]]
b = copy.deepcopy(a)
a[1].append(30)
print(b)
```

Explanation:
deepcopy duplicates all nested objects, so b has its own inner lists separate from a.
When you append 30 to a[1], it changes only a’s inner list, not b’s.

```python
# Q39
import copy

a = [[1, 2], [3, 4]]
b = copy.copy(a)
print(a is b)
print(a[0] is b[0])
```

Explanation:
copy.copy() makes a shallow copy — it creates a new outer container but does not copy nested objects.
So a and b are different lists (is → False), but their inner lists are shared (is → True).
