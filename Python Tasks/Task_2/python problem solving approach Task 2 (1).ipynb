Q1. Find the Shortest Word in a Sentence
Six-Step Problem-Solving Approach
Observe: The input is a sentence and the output should be its shortest word. Understand: Compare the lengths of all words. Break: Split the sentence into words, then track the shortest word. Logic: Start with the first word and replace it only when a strictly shorter word is found; this preserves the first word when lengths tie. Code: Implement the comparison using a loop without relying on min(). Test: Run the given example and an additional sentence.

Given input: Python makes coding easy
Assignment expected output: easy

```python
def shortest_word(sentence):
    words = sentence.split()
    if not words:
        return None
    shortest = words[0]
    for word in words[1:]:
        if len(word) < len(shortest):
            shortest = word
    return shortest

print(shortest_word("Python makes coding easy"))
```

Q6. Count Vowels in a String

```python
def count_vowels(text):
    vowels = "aeiou"
    count = 0
    for char in text:
        if char.lower() in vowels:
            count += 1
    return count

print(count_vowels("Education"))
```

Q7. Count Consonants in a String

Six-Step Problem-Solving Approach

Break: Inspect each character, keep alphabetic characters, and count letters that are not vowels. Logic: Convert letters to lowercase and test against the vowel set. Code: Use isalpha() and a vowel check. Test: Run the supplied input and an additional input containing punctuation.

Given input: Python 123!

output: 5

```python
def count_consonants(text):
    vowels = "aeiou"
    count = 0
    for char in text:
        if char.isalpha() and char.lower() not in vowels:
            count += 1
    return count

print(count_consonants("Python 123!"))
```

Q9. Find the Least Frequent Character

Six-Step Problem-Solving Approach
Observe: The output is the least frequent non-space character. Understand: Ties must be resolved by first appearance. Break: Count characters, find the minimum frequency, then scan the original string. Logic: The first character whose frequency equals the minimum is the answer. Code: Use a frequency dictionary and preserve input order when selecting. Test: Run the supplied example and another string.

Given input: success

output: u

```python
def least_frequent_character(text):
    counts = {}
    for char in text:
        if char != " ":
            counts[char] = counts.get(char, 0) + 1
    if not counts:
        return None
    minimum = min(counts.values())
    for char in text:
        if char != " " and counts[char] == minimum:
            return char
    return None

print(least_frequent_character("success"))
```

Q10. Find All Characters Appearing More Than Once

Six-Step Problem-Solving Approach
Observe: The output is a list of characters occurring more than once, with spaces ignored. Understand: Each repeated character should appear only once and in first-appearance order. Break: Count characters, then scan the original string and collect characters whose counts exceed one. Logic: Use a dictionary for counts and a result list to preserve order. Code: Add a character only if it repeats and has not already been added. Test: Run the supplied example and an additional string.

Given input: programming
    
Assignment expected output: ['r', 'g', 'm']

```python
def repeated_characters(text):
    counts = {}
    for char in text:
        if char != " ":
            counts[char] = counts.get(char, 0) + 1
    result = []
    for char in text:
        if char != " " and counts[char] > 1 and char not in result:
            result.append(char)
    return result

print(repeated_characters("programming"))
```

Q16. Find the Most Frequent Word in a Sentence

Six-Step Problem-Solving Approach
Observe: The output is the most frequent word, ignoring case. Understand: Uppercase and lowercase forms of a word are considered the same. Break: Convert words to lowercase, count frequencies, then preserve first appearance for ties. Logic: Build a frequency dictionary and scan the normalized words in order. Code: Use lower() and a dictionary. Test: Run the supplied example and another example with mixed case.

Given input: Python is easy and Python is powerful

Assignment expected output: python

```python
def most_frequent_word(sentence):
    words = sentence.lower().split()
    if not words:
        return None
    counts = {}
    for word in words:
        counts[word] = counts.get(word, 0) + 1
    best = words[0]
    for word in words[1:]:
        if counts[word] > counts[best]:
            best = word
    return best

print(most_frequent_word("Python is easy and Python is powerful"))
```

Q18. Find the First Word That Appears Only Once

Six-Step Problem-Solving Approach
Observe: The output is the first word whose frequency is exactly one. Understand: Matching is case-insensitive. Break: Normalize words, count them, then scan in original order. Logic: Return the first normalized word with frequency one. Code: Use a frequency dictionary followed by an ordered scan. Test: Run the supplied sentence and an additional sentence.

Given input: apple banana apple mango banana orange

output: mango



```python
def first_unique_word(sentence):
    words = sentence.lower().split()
    counts = {}
    for word in words:
        counts[word] = counts.get(word, 0) + 1
    for word in words:
        if counts[word] == 1:
            return word
    return None

print(first_unique_word("apple banana apple mango banana orange"))
```

Q20. Find the Longest Word That Appears Only Once

Six-Step Problem-Solving Approach
Observe: The output is the longest word whose frequency is exactly one. Understand: Matching is case-insensitive, and ties among qualifying words return the first one encountered. Break: Normalize words, count frequencies, then scan the sentence while tracking the longest unique word. Logic: Consider only words with count one and update the result only for a strictly longer word. Code: Use a frequency dictionary and an ordered scan. Test: Run the supplied example and another example.

Given input: Python makes coding useful Python

output: coding

```python
def longest_unique_word(sentence):
    words = sentence.lower().split()
    counts = {}
    for word in words:
        counts[word] = counts.get(word, 0) + 1
    result = None
    for word in words:
        if counts[word] == 1:
            if result is None or len(word) > len(result):
                result = word
    return result

print(longest_unique_word("Python makes coding useful Python"))
```
